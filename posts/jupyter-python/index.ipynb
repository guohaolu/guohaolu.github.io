下面执行一段 Python，并用 Matplotlib 画图。


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi, 200)
y = np.sin(x)

fig, ax = plt.subplots(figsize=(7, 3.5))
ax.plot(x, y, color="#2c7fb8", linewidth=2)
ax.set_title("y = sin(x)")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.grid(True, alpha=0.3)
plt.show()
```
